```python
# Mini Q3 experiment — only mass vs log_mass, with and without scaling

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# ---------------------------
# Load data
# ---------------------------
df = pd.read_csv("exoplanet.csv")

# ensure numeric
df["pl_bmasse"] = pd.to_numeric(df["pl_bmasse"], errors="coerce")
df = df.dropna(subset=["pl_bmasse", "planet_class"])

# create log mass safely
df["pl_bmasse"] = df["pl_bmasse"].replace(0, np.nan)
df["log_pl_bmasse"] = np.log(df["pl_bmasse"])

# ---------------------------
# Function to train + report
# ---------------------------
def test_feature(column_name, scale=False):
    print("\n==============================")
    print(f"TESTING FEATURE: {column_name} | scaling={scale}")
    print("==============================")

    X = df[[column_name]].fillna(df[column_name].median()).values
    y = df["planet_class"].astype(str).values

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.25, stratify=y, random_state=42
    )

    if scale:
        pipe = Pipeline([
            ("scaler", StandardScaler()),
            ("svc", SVC(kernel="linear"))
        ])
        pipe.fit(X_train, y_train)
        preds = pipe.predict(X_test)
    else:
        model = SVC(kernel="linear")
        model.fit(X_train, y_train)
        preds = model.predict(X_test)

    print("Accuracy:", round(accuracy_score(y_test, preds), 4))
    print(classification_report(y_test, preds))

# ---------------------------
# Run 4 tests
# ---------------------------

# 1) only pl_bmasse, no scaling
test_feature("pl_bmasse", scale=False)

# 2) only pl_bmasse, WITH scaling
test_feature("pl_bmasse", scale=True)

# 3) only log_pl_bmasse, no scaling
test_feature("log_pl_bmasse", scale=False)

# 4) only log_pl_bmasse, WITH scaling
test_feature("log_pl_bmasse", scale=True)
```
